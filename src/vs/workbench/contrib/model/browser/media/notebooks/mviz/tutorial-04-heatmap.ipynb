# Heatmap

A heatmap shows a 2-D matrix as a colour grid — ideal for correlation matrices and any dense array. Colour scale choice is critical.

```julia
Pollis.packages("Plots")
```

```julia
using Plots
using Statistics
using Random

Random.seed!(23);
```

## Data Matrix Heatmap

A plain value matrix with a sequential colour scale for non-negative magnitudes.

```julia
M = rand(10, 12)
heatmap(M; color = :viridis, xlabel = "Column", ylabel = "Row", title = "Data Matrix", colorbar_title = "Value")
```

## Correlation Heatmap

For signed data like correlations, use a **diverging** scale centred at zero so colour direction encodes sign. The all-ones diagonal carries no information — read the off-diagonal cells.

```julia
# five variables with some correlation structure
a = randn(300)
X = hcat(a, 0.8 .* a .+ randn(300), -0.5 .* a .+ randn(300), randn(300), 0.3 .* a .+ randn(300))
C = cor(X)

heatmap(C; color = :RdBu, clims = (-1, 1), title = "Correlation Matrix",
    xticks = (1:5, ["v1","v2","v3","v4","v5"]), yticks = (1:5, ["v1","v2","v3","v4","v5"]), colorbar_title = "r")
```

`clims = (-1, 1)` fixes the scale so zero sits at the neutral midpoint. v1 and v2 are strongly positive (deep blue/red per palette), v1 and v3 negative.

## Summary
- Sequential palette for magnitudes, diverging palette for signed data.
- Centre a diverging scale at zero (`clims`).
- Always show the colour bar; ignore the trivial correlation diagonal.
