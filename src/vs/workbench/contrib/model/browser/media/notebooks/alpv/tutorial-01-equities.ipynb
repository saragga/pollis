# Equity Time Series

Download equity OHLC data from Alpha Vantage with [AlphaVantage.jl](https://github.com/ellisvalentiner/AlphaVantage.jl).

**API key:** AlphaVantage.jl reads `ENV["ALPHA_VANTAGE_API_KEY"]`. Set it for this session before the first call (or use the webview's *Set API key* link, which injects it into the Julia REPL).

```julia
# ENV["ALPHA_VANTAGE_API_KEY"] = "YOURKEY"  # uncomment and set if not already in the environment
using AlphaVantage
```

## Daily and Intraday Bars

```julia
daily = time_series_daily("AAPL"; outputsize="compact")
println(daily)
```

```julia
intraday = time_series_intraday("AAPL"; interval="5min")
println(intraday)
```

## Latest Global Quote

```julia
quote = stock_quote("AAPL")
println(quote)
```

## Summary
- `time_series_daily` / `_intraday` / `_weekly` / `_monthly` return OHLC series.
- `outputsize="full"` returns 20+ years; `"compact"` the latest 100 points.
- `stock_quote` returns a single latest snapshot.
