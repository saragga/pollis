# Choropleth

A choropleth shades geographic regions by a value using a colour scale. This tutorial uses [GeoMakie.jl](https://github.com/MakieOrg/GeoMakie.jl) for the map axis and [NaturalEarth.jl](https://github.com/JuliaGeo/NaturalEarth.jl) for region polygons.

```julia
Pollis.packages("GeoMakie", "CairoMakie", "NaturalEarth")
```

```julia
using GeoMakie
using CairoMakie
using NaturalEarth
using Random

Random.seed!(42);
```

## Region Polygons

`naturalearth` returns a GeoTable of country polygons at the requested resolution (110 m is low-res, fast to draw).

```julia
countries = naturalearth("admin_0_countries", 110)
n = length(countries.geometry)
values = rand(n);     # one value per region — replace with your data joined on a key
```

## Draw the Map

A `GeoAxis` applies a map projection. `poly!` shades each region by its value; a `Colorbar` documents the scale.

```julia
fig = Figure(; size = (900, 450))
ga  = GeoAxis(fig[1, 1]; title = "Choropleth")
poly!(ga, countries.geometry; color = values, colormap = :viridis, strokecolor = :white, strokewidth = 0.25)
Colorbar(fig[1, 2]; colormap = :viridis, label = "Value")
fig
```

**Colour is the value** — always read the colour bar. Watch for two traps: large regions dominate the eye regardless of value, and the way values are classified into colours changes the apparent pattern.

## Use Rates, Not Counts

For counts (population, sales), shade a **rate** (per-capita / per-area) instead, or the map mostly shows where people are, not the variable of interest.

```julia
area = rand(n) .+ 0.5            # stand-in for region area or population
rate = values ./ area            # normalise
fig2 = Figure(; size = (900, 450))
ga2  = GeoAxis(fig2[1, 1]; title = "Rate (value / area)")
poly!(ga2, countries.geometry; color = rate, colormap = :viridis, strokewidth = 0.25)
Colorbar(fig2[1, 2]; colormap = :viridis, label = "Rate")
fig2
```

## Summary
- Choropleth = colour-on-map; consult the colour bar.
- Normalise to rates so the map shows the variable, not population/area.
- State the classification (quantile vs equal-interval) and use a sensible projection.
