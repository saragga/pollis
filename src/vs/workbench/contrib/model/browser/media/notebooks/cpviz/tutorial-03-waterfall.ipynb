# Waterfall Plot

A waterfall plot shows how a starting value becomes a final value through a sequence of signed contributions, drawn as **floating bars** that step up (gains) and down (losses).

```julia
using Plots
```

## Steps and Deltas

An ordered list of signed changes with a defined start. The final bar is the running total.

```julia
labels = ["Start", "+Sales", "-Costs", "+Other", "End"]
deltas = [100, 45, -30, 20]      # signed change at each step
cum    = cumsum(deltas);         # running total after each step
```

## Floating Bars

Each intermediate bar hangs between the running total before and after its step. The `fillto` keyword sets each bar's base; the first and last bars are anchored to zero.

```julia
tops    = [cum; cum[end]]        # top of each bar (End = final total)
bottoms = [0; cum[1:end-1]; 0]   # base of each bar

bar(labels, tops; fillto = bottoms, legend = false,
    title = "Waterfall", ylabel = "Value", xrotation = 20)
```

## Reading It

Bars rising from the previous level are increases; bars dropping are decreases. The final bar equals the cumulative sum of all deltas — that is the bridge from start to end the chart exists to show.

```julia
println("Final total: ", cum[end])
```

## Summary
- Floating bars step up and down by signed deltas.
- `fillto` sets each bar base; start and end anchor to zero.
- End bar = cumulative sum — a mismatch signals a sign or data error.
