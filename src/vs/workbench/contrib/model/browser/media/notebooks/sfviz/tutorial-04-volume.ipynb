# Volume

A volume plot visualises a full 3D scalar field `f(x, y, z)` — as nested **isosurfaces** or via **direct volume rendering**, which accumulates the field along each viewing ray so dense regions look brighter. It reveals internal structure no surface can. Requires [GLMakie](https://github.com/MakieOrg/Makie.jl) (GPU/display).

```julia
Pollis.packages("GLMakie")
```

```julia
using GLMakie
```

## A 3D Field

`cube[i, j, k] = f(xs[i], ys[j], zs[k])`. Keep the resolution modest — a 3D array grows as the cube.

```julia
r    = LinRange(-1, 1, 80)
cube = [x^2 + y^2 + z^2 for x in r, y in r, z in r];
```

## Nested Isosurfaces

`contour` on a 3D array draws shells of constant value; transparency lets you see inside.

```julia
contour(cube; alpha = 0.5)
```

Each shell is a surface of constant value; their spacing shows how quickly the field changes through the cube. Keep some **transparency** (`alpha`), or the outer shells hide the interior. Rotate the GLMakie window to see inside.

(`volume(cube; algorithm = :mip)` is the direct ray-integrated alternative — a soft cloud rather than crisp shells.)

## Summary
- A volume plot shows `f(x, y, z)` across a 3D region.
- `contour(cube; alpha = ...)` draws nested isosurface shells; keep some transparency so the interior shows.
- GLMakie only (opens a window); keep grids modest (~50-100 per axis).
