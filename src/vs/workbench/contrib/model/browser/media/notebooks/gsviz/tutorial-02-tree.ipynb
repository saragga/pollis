# Tree Diagram

A tree diagram lays out a hierarchy top-down so parent-child structure reads at a glance. This tutorial uses the **Buchheim** tidy-tree algorithm from [NetworkLayout.jl](https://github.com/JuliaGraphs/NetworkLayout.jl) with [GraphMakie.jl](https://github.com/MakieOrg/GraphMakie.jl).

```julia
Pollis.packages("Graphs", "GraphMakie", "NetworkLayout", "CairoMakie")
```

```julia
using Graphs
using GraphMakie
using NetworkLayout
using CairoMakie
```

## A Rooted Tree

`binary_tree(4)` is an **undirected** graph, but Buchheim needs a **rooted tree** — directed edges from each parent to its children — so it can find the root and lay out depth. `bfs_tree(g, 1)` directs the edges away from node 1, turning the undirected tree into a rooted one.

```julia
ug = binary_tree(4)        # undirected complete binary tree of depth 4
g  = bfs_tree(ug, 1)       # direct edges away from root 1 -> a rooted tree
```

## Tidy Layout

The Buchheim algorithm produces a compact, non-overlapping layered layout: the root at the top, depth increasing downward, children below their parent.

```julia
graphplot(g;
    layout  = Buchheim(),
    nlabels = string.(1:nv(g)),
    node_size = 12)
```

Trace any path root → leaf to follow the hierarchy; compare subtree sizes to spot imbalance or unusually deep branches.

## Building Your Own Hierarchy

Add edges parent → child to encode a real structure (e.g. an org chart or taxonomy).

```julia
h = SimpleDiGraph(7)
# 1 = CEO; 2,3 = VPs; 4..7 = teams
add_edge!.(Ref(h), [1, 1, 2, 2, 3, 3], [2, 3, 4, 5, 6, 7])
graphplot(h; layout = Buchheim(), nlabels = ["CEO","VP-A","VP-B","T1","T2","T3","T4"])
```

## Summary
- A tree is a hierarchy: one parent per node, a single root.
- Buchheim gives a clean, compact top-down layout.
- For general (cyclic, many-to-many) relationships use a Network Diagram instead.
