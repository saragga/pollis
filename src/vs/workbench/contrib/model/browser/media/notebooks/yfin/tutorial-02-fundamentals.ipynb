# Fundamentals

Pull financial statements and valuation multiples from Yahoo Finance with [YFinance.jl](https://github.com/eohne/YFinance.jl).

```julia
using YFinance, Dates
```

## Income Statement

`get_Fundamental(symbol, item, interval, startdt, enddt)` returns one column per line item plus a `timestamp` of period-end dates. Valid statements: `"income_statement"`, `"balance_sheet"`, `"cash_flow"`, `"valuation"`.

```julia
inc = get_Fundamental("AMD", "income_statement", "annual", today() - Year(3), today())
println("Line items: ", keys(inc))
println("Total revenue: ", inc["TotalRevenue"])
```

## Valuation Multiples

The `"valuation"` item exposes ratios such as PE, PB, PEG, and EV/EBITDA.

```julia
val = get_Fundamental("NFLX", "valuation", "annual", today() - Year(3), today())
println("Periods:   ", val["timestamp"])
println("PE ratio:  ", val["PeRatio"])
println("PEG ratio: ", val["PegRatio"])
```
