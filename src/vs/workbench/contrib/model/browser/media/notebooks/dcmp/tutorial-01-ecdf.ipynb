# ECDF and CDF

The empirical cumulative distribution function (ECDF) is an assumption-free picture of a distribution. Overlaying a theoretical CDF gives a direct goodness-of-fit comparison.

```julia
using StatsPlots
using Distributions
using Random
using Statistics

Random.seed!(10);
```

## Empirical CDF

The ECDF steps up by 1/n at each observation.

```julia
x = randn(200)
ecdfplot(x; xlabel = "Value", ylabel = "F(x)", label = "Empirical", title = "ECDF")
```

## Overlay a Theoretical CDF

Fit a Normal to the sample mean and standard deviation, then overlay its CDF. The largest vertical gap is the Kolmogorov-Smirnov distance.

```julia
ecdfplot(x; label = "Empirical", xlabel = "Value", ylabel = "F(x)")
plot!(t -> cdf(Normal(mean(x), std(x)), t); lw = 2, label = "Normal CDF")
```

## Comparing Two Groups

Two ECDFs on one axis compare distributions directly — a horizontal shift is a location difference, different steepness is a spread difference.

```julia
y = 0.8 .* randn(200) .+ 1.0
ecdfplot(x; label = "Group A", xlabel = "Value", ylabel = "F(x)")
ecdfplot!(y; label = "Group B")
```

## Summary
- The ECDF needs no binning or bandwidth.
- The biggest empirical-vs-theoretical gap is the KS statistic.
- Two ECDFs compare groups across the whole distribution, not just the mean.
