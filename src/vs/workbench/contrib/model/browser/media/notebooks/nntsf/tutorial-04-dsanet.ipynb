# DSANet — Neural Network Time-Series Forecasting

Dual Self-Attention Network (Huang et al., 2019): global and local self-attention branches with no recurrence, for many correlated series.

**Coming soon!** This is a starter stub — run the cell below to train the model on the bundled sample data.

```julia
using FluxArchitectures

poollength = 10
input, target = get_data(:exchange_rate, poollength, 1000, 6)
inputsize = size(input, 1)

model = DSANet(inputsize, poollength, 3, 3, 4, 1, 1, 1, 6)
loss(x, y) = (Flux.reset!(model); Flux.mse(model(x), transpose(y)))
opt = Flux.ADAM(0.007)
Flux.train!(loss, Flux.params(model), Iterators.repeated((input, target), 20), opt)
```
