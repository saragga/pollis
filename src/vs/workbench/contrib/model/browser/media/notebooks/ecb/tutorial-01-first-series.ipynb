# Your First ECB Series

Download a series from the [ECB Data Portal](https://data.ecb.europa.eu) into Julia. The API is a standard SDMX REST service and needs **no API key** — we ask for `format=csvdata` and read the flat table with `HTTP.jl` + `CSV.jl` + `Tables.jl`.

```julia
# import Pkg; Pkg.add(["HTTP", "CSV", "Tables"])
using HTTP, CSV, Tables

base = "https://data-api.ecb.europa.eu/service/data"
```

## Fetch the US Dollar / Euro Reference Rate

The dataflow is `EXR` and the key is `D.USD.EUR.SP00.A` (daily, USD per EUR, reference rate, average).

```julia
url = "$base/EXR/D.USD.EUR.SP00.A?startPeriod=2015-01-01&format=csvdata"
resp = HTTP.get(url)
tbl  = CSV.File(resp.body)        # Tables.jl-compatible
```

## Read the Columns

The CSV is flat, so the columns we want are `TIME_PERIOD` and `OBS_VALUE`. `Tables.getcolumn` works on any Tables.jl source.

```julia
dates  = Tables.getcolumn(tbl, :TIME_PERIOD)
values = Tables.getcolumn(tbl, :OBS_VALUE)

println("Observations: ", length(values))
println(last(dates, 5))
println(last(values, 5))
```

## Summary
- The ECB Data Portal serves SDMX over REST at `data-api.ecb.europa.eu/service/`, with no key.
- `format=csvdata` returns a flat CSV that `CSV.File` reads into a Tables.jl table.
- Pull `TIME_PERIOD` / `OBS_VALUE` with `Tables.getcolumn`, or pipe the table into TimeSeries or any other Tables.jl sink.
