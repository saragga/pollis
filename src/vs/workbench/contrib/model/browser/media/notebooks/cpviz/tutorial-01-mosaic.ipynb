# Mosaic Plot

A mosaic plot tiles a unit square so that each cell's **area** is proportional to its frequency in a two-way contingency table. It is the natural way to *see* the association between two categorical variables, and pairs with a chi-squared test.

```julia
using Plots
using HypothesisTests
```

## A Contingency Table

Rows are the levels of variable X, columns the levels of variable Y. Here: three regions (rows) by two outcomes (columns).

```julia
counts = [40 25;   # Region A
          18 30;   # Region B
          12 20]   # Region C
labels_x = ["A", "B", "C"]
labels_y = ["No", "Yes"];
```

## Building the Mosaic

Column **widths** come from the column totals (the marginal of Y); within each column the **row split** is the conditional distribution of X. We draw one `Shape` rectangle per cell.

```julia
coltot = vec(sum(counts; dims = 1))
xedges = [0; cumsum(coltot ./ sum(coltot))]   # column boundaries on [0, 1]

plt = plot(; legend = false, xlims = (0, 1), ylims = (0, 1),
    title = "Mosaic Plot", xlabel = "Y (width)", ylabel = "X (height)")

for j in 1:size(counts, 2)
    yfrac  = counts[:, j] ./ sum(counts[:, j])   # row split within this column
    yedges = [0; cumsum(yfrac)]
    for i in 1:size(counts, 1)
        plot!(plt, Shape([xedges[j], xedges[j+1], xedges[j+1], xedges[j]],
                         [yedges[i], yedges[i], yedges[i+1], yedges[i+1]]);
              fillalpha = 0.25 + 0.2i, c = i)
    end
end
plt
```

Cell area is proportional to frequency. If the row splits look the same across columns, X and Y are independent; if they shift, the variables are associated.

## Testing the Association

A Pearson chi-squared test confirms whether an apparent association is statistically significant.

```julia
test = ChisqTest(counts)
println("chi2 = ", round(test.stat; digits = 3), "  p = ", pvalue(test))
# p < 0.05 suggests the two categorical variables are associated
```

## Summary
- Mosaic tile **area** encodes contingency-table frequency.
- Column width = marginal of one variable; row split = conditional of the other.
- Differing splits across columns indicate association — confirm with a chi-squared test.
