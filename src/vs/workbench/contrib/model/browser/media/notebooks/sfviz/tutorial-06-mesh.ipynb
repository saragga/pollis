# Mesh

A mesh renders an explicit surface from **vertices** and the **faces** that connect them — the general form behind every 3D surface. Uses [GLMakie](https://github.com/MakieOrg/Makie.jl) with [GeometryBasics](https://github.com/JuliaGeometry/GeometryBasics.jl) (CairoMakie also renders a static mesh).

```julia
using GLMakie, GeometryBasics, LinearAlgebra
```

## Vertices on a Sphere Grid

Sample a sphere on a `(theta, phi)` grid, then flatten the grid into a flat list of `Point3f` vertices.

```julia
r = 0.5f0; n = 30
theta = LinRange(0, pi, n)
phi   = LinRange(0, 2pi, 2n)
x = [r*cos(p)*sin(t) for t in theta, p in phi]
y = [r*sin(p)*sin(t) for t in theta, p in phi]
z = [r*cos(t)        for t in theta, p in phi]
points = vec([Point3f(xv, yv, zv) for (xv, yv, zv) in zip(x, y, z)]);
```

## Faces and Normals

Neighbouring grid vertices form quad faces. For a centred sphere the outward normal at each vertex is just the (normalised) vertex itself, which gives smooth shading.

```julia
faces   = decompose(QuadFace{GLIndex}, Tessellation(Rect(0, 0, 1, 1), size(z)))
gb_mesh = GeometryBasics.Mesh(points, faces; normal = Vec3f.(normalize.(points)));
```

## Mesh and Wireframe

`mesh` draws the filled surface; a `wireframe!` overlay exposes the underlying quad grid.

```julia
f, ax, pl = mesh(gb_mesh; color = [v[3] for v in points], colormap = :blues)
wireframe!(ax, gb_mesh; color = (:black, 0.2), linewidth = 1)
f
```

## Summary
- A mesh is vertices + faces; normals drive the shading.
- `decompose(QuadFace{GLIndex}, Tessellation(...))` connects a grid into quads.
- Colour per-vertex (length = number of vertices); overlay `wireframe!` to show topology.
