# Direct Methods: Trajectory Optimization via Collocation

This tutorial demonstrates direct collocation for trajectory optimization using TrajectoryOptimization.jl and Ipopt.

```julia
using TrajectoryOptimization, Ipopt, Plots, LinearAlgebra

# Define simple 2D linear dynamics
A = [0 1; -1 0.5]
B = [0; 1]

n, m = 2, 1  # state and control dimensions
T = 10       # horizon
h = 0.1      # time step

# Create problem
prob = Problem(A, B, n, m, T, h)

# Set cost function: Q weight on states, R weight on controls
Q = 1.0 * I(n)
R = 0.1 * I(m)

# Set boundary conditions
x0 = [1; 0]
xf = [0; 0]

# Solve
solve!(prob)
```

## Results

Extract and visualize the optimal trajectory and control inputs.

```julia
# Get solution
X = states(prob)
U = controls(prob)

# Plot trajectory
plot(X[1,:], X[2,:], label="State trajectory", xlabel="x1", ylabel="x2")

# Plot controls
plot(U[1,:], label="Control u(t)")
```
