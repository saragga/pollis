# ALTRO: Augmented Lagrangian Trajectory Optimizer

State-of-the-art constrained trajectory optimization combining DDP with augmented Lagrangian framework.

```julia
Pollis.packages("Altro", "TrajectoryOptimization", "Plots")
```

```julia
using Altro, TrajectoryOptimization, LinearAlgebra, Plots

# Define constrained trajectory problem
# dynamics: dx = Ax + Bu
# constraints: state bounds, control bounds, path constraints

n, m = 3, 2
T = 50

# Create ALTRO problem
prob = Problem(TrajectoryOptimization.DoubleIntegrator(n))

# Add constraints
# Control limits: ||u|| <= u_max
u_max = 1.0

# State constraints: x in box [x_min, x_max]
x_min = [-2, -2, -2]
x_max = [2, 2, 2]

# Set up ALTRO solver
solver = ALTROSolver(prob)
set_options!(solver, show_summary=true, iterations=100)

# Solve
solve!(solver)
```

## Results and Feasibility Check

Verify constraint satisfaction and analyze solution.

```julia
# Extract solution
X = states(solver)
U = controls(solver)

# Verify constraints
for i = 1:length(X)
    @assert all(X[i] .<= x_max) && all(X[i] .>= x_min)
end

for i = 1:length(U)
    @assert norm(U[i]) <= u_max + 1e-6
end

println("All constraints satisfied!")

# Plot
plot(X[1,:], label="x1")
plot!(X[2,:], label="x2")
```
