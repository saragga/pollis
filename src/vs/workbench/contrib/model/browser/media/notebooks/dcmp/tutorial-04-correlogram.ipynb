# Correlogram

A correlogram (scatter-plot matrix) lays out pairwise scatters of every variable combination with marginal densities on the diagonal — a fast survey of multivariate structure.

```julia
Pollis.packages("StatsPlots")
```

```julia
using StatsPlots
using Random

Random.seed!(13);
```

## Build a Small Multivariate Dataset

Three correlated variables plus one independent one.

```julia
n = 300
a = randn(n)
b = 0.8 .* a .+ 0.6 .* randn(n)
c = -0.5 .* a .+ 0.9 .* randn(n)
d = randn(n)
M = hcat(a, b, c, d);
```

## Corner Plot

`corrplot` draws the scatter matrix; tilted off-diagonal clouds indicate correlation, the diagonal shows each marginal.

```julia
corrplot(M; label = ["a" "b" "c" "d"], size = (700, 700))
```

- `a`–`b` panel: tilted up → positive correlation.
- `a`–`c` panel: tilted down → negative correlation.
- `d` panels: shapeless → independent.

## Drilling In

Use the correlogram to spot interesting pairs, then inspect one closely with a dedicated scatter.

```julia
scatter(a, b; xlabel = "a", ylabel = "b", title = "a vs b", alpha = 0.5, legend = false)
```

## Summary
- Off-diagonal panels → pairwise relationships; diagonal → marginals.
- Best for ~6-8 variables; beyond that it gets cramped.
- Use it to find pairs worth a closer look.
