# Your First Fama-French Factors

Download factor returns from Kenneth French's Data Library with `FamaFrenchData.jl`. There is **no API key**. Each file comes back as a vector of `Tables.jl` tables plus their notes.

```julia
Pollis.packages("FamaFrenchData", "Tables", "Plots")
```

```julia
# import Pkg; Pkg.add(["FamaFrenchData", "Tables", "Plots"])
using FamaFrenchData, Tables, Plots, Dates

tables, tablenotes, filenotes = readFamaFrench("F-F_Research_Data_Factors")
tablenotes
```

## The Monthly Factors

Table 1 holds the monthly factors, in percent. Dates are integers such as `192607`; `Mkt-RF` needs `Symbol` because of the hyphen.

```julia
monthly = tables[1]
date = Tables.getcolumn(monthly, :Date)
mktrf = Tables.getcolumn(monthly, Symbol("Mkt-RF"))
smb = Tables.getcolumn(monthly, :SMB)
hml = Tables.getcolumn(monthly, :HML)
rf = Tables.getcolumn(monthly, :RF)
dates = Date.(div.(date, 100), rem.(date, 100))
println(first(dates), " to ", last(dates), ", ", length(dates), " months")
```

## Average Premiums and Sharpe Ratios

Annualise the monthly mean by 12 and the standard deviation by the square root of 12.

```julia
using Statistics

for (name, r) in ["Mkt-RF" => mktrf, "SMB" => smb, "HML" => hml]
    println(rpad(name, 8), "mean ", round(12 * mean(r); digits = 2), "% a year,  Sharpe ", round(sqrt(12) * mean(r) / std(r); digits = 2))
end
```

## Growth of One Dollar

Each factor is a long-short return, so its cumulative product shows how a strategy that earns it would have grown. The vertical axis is logarithmic.

```julia
grow(r) = cumprod(1 .+ r ./ 100)
plot(dates, [grow(mktrf) grow(smb) grow(hml)], yscale = :log10,
     label = ["Mkt-RF" "SMB" "HML"], xlabel = "date", ylabel = "value of 1 dollar", legend = :topleft)
```
