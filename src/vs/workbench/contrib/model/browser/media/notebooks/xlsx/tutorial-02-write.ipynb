# Write and Format

This tutorial creates a styled multi-sheet workbook from Julia DataFrames. Writing uses `XLSX.writetable`; formatting uses `XLSX.openxlsx` with `setFont`, `setFill`, `setBorder`, and `setAlignment` (XLSX.jl v0.11+).

```julia
using XLSX
using DataFrames
```

## Prepare Data

```julia
df_sales = DataFrame(
    product = ["Widget A", "Widget B", "Widget C"],
    units   = [120, 85, 200],
    revenue = [2400.0, 1700.0, 4000.0],
)

df_costs = DataFrame(
    product = ["Widget A", "Widget B", "Widget C"],
    cogs    = [800.0, 600.0, 1200.0],
)
```

## Write Two Sheets

```julia
XLSX.writetable("report.xlsx",
    "Sales" => collect(eachcol(df_sales)),
    "Costs" => collect(eachcol(df_costs)))
```

## Apply Formatting

Open the file in read-write mode and style the header row of the Sales sheet.

```julia
XLSX.openxlsx("report.xlsx"; mode="rw") do xf
    ws = xf["Sales"]

    for col in ["A", "B", "C"]
        cell = col * "1"
        XLSX.setFont(ws, cell; bold=true, color="FFFFFF", size=11)
        XLSX.setFill(ws, cell; fgColor="4472C4")
        XLSX.setAlignment(ws, cell; horizontal="center")
    end

    # Highlight revenue column: green if > 2000
    XLSX.setConditionalFormat(ws, "C2:C100";
        rule=">2000", fgColor="C6EFCE")
end
```

## Summary
- `writetable` writes multiple sheets in one call — fastest path for plain data.
- `openxlsx` with `mode="rw"` opens an existing file for formatting.
- `setFont`, `setFill`, `setAlignment` style individual cells or ranges.
- `setConditionalFormat` adds rule-based highlighting evaluated by Excel.
