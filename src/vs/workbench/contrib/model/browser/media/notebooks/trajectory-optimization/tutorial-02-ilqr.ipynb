# Indirect Methods: Iterative Linear Quadratic Regulator (iLQR)

This tutorial demonstrates iLQR for local trajectory refinement via successive quadratic approximation.

```julia
Pollis.packages("TrajectoryOptimization", "Plots")
```

```julia
using TrajectoryOptimization, LinearAlgebra, Plots

# Nonlinear pendulum dynamics
function f(x, u)
    [x[2]; -sin(x[1]) + u[1]]
end

n, m = 2, 1
T = 50
dt = 0.1

# Cost: ||x_f|| + integral of ||x||^2 + ||u||^2
# Initial trajectory
x_traj = [zeros(n) for _ = 1:T]
u_traj = [zeros(m) for _ = 1:T-1]

# Initialize with small random perturbation
x_traj[1] = [pi; 0]  # Start at bottom

# iLQR iteration
for iter = 1:10
    # Forward dynamics
    for k = 1:T-1
        x_traj[k+1] = x_traj[k] + dt * f(x_traj[k], u_traj[k])
    end
    
    # Backward pass: compute gains
    # ... (simplified; full backward pass computes K(k) gains)
    
    println("Iteration $iter: cost = ", sum(norm.(x_traj)) + sum(norm.(u_traj)))
end
```

## Convergence Analysis

Monitor convergence to ensure algorithm succeeds.

```julia
# Plot trajectory
plot([x[1] for x in x_traj], [x[2] for x in x_traj], label="Phase portrait")
plot!([u[1] for u in u_traj], label="Controls")
```
