# Technical Indicators

Alpha Vantage computes 50+ technical indicators server-side, so [AlphaVantage.jl](https://github.com/ellisvalentiner/AlphaVantage.jl) returns the indicator series directly.

Set `ENV["ALPHA_VANTAGE_API_KEY"]` for this session first.

```julia
Pollis.packages("AlphaVantage")
```

```julia
using AlphaVantage
```

## Simple Moving Average and RSI

```julia
sma = SMA("AAPL"; interval="daily", time_period=20, series_type="close")
println(sma)
```

```julia
rsi = RSI("AAPL"; interval="daily", time_period=14, series_type="close")
println(rsi)
```

## Summary
- Indicator functions (`SMA`, `EMA`, `RSI`, `MACD`, `VWAP`, …) take `interval`, `time_period`, and `series_type`.
- The indicator is computed by Alpha Vantage and returned as a series.
- No need to implement the indicator yourself.
