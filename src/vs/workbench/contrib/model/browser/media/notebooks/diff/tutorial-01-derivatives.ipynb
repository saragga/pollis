# Derivatives and Tangent Lines

The derivative $f'(x_0)$ is the slope of the tangent to $f$ at $x_0$: the rate at which $f$ changes. In economics it is a *marginal* quantity. This notebook computes first and second derivatives with DifferentiationInterface and ForwardDiff and reads the shape of a function from them.

```julia
using DifferentiationInterface, Plots
import ForwardDiff
backend = AutoForwardDiff()
```

## A cost function

Take total cost $C(q) = 50 + 10q - 0.8q^2 + 0.1q^3$. Marginal cost is $C'(q)$; whether marginal cost is rising is $C''(q)$.

```julia
C(q) = 50 + 10q - 0.8q^2 + 0.1q^3
q0 = 3.0
(marginal_cost = derivative(C, backend, q0), slope_of_mc = second_derivative(C, backend, q0))
```

Check by hand: $C'(q) = 10 - 1.6q + 0.3q^2$ and $C''(q) = -1.6 + 0.6q$. At $q = 3$ these are $7.9$ and $0.2$.

## The tangent line and the quadratic approximation

$C(q_0 + \Delta q) \approx C(q_0) + C'(q_0)\Delta q$ (tangent) and adding $\tfrac12 C''(q_0)\Delta q^2$ gives the quadratic approximation.

```julia
d1 = derivative(C, backend, q0); d2 = second_derivative(C, backend, q0)
qs = range(0, 8, length=300)
plot(qs, C.(qs), lw=2, c=:black, label="C(q)", xlabel="q")
plot!(qs, C(q0) .+ d1 .* (qs .- q0), lw=2, c=:blue, label="tangent at q0")
plot!(qs, C(q0) .+ d1 .* (qs .- q0) .+ d2 / 2 .* (qs .- q0) .^ 2, lw=2, ls=:dash, c=:red, label="quadratic")
scatter!([q0], [C(q0)], c=:black, label="")
```

## Marginal and average cost

Average cost $C(q)/q$ is lowest where it equals marginal cost. Plot both curves using the derivative at every point; `prepare_derivative` does the setup once.

```julia
prep = prepare_derivative(C, backend, q0)
MC = [derivative(C, prep, backend, q) for q in qs[2:end]]
AC = C.(qs[2:end]) ./ qs[2:end]
plot(qs[2:end], [MC AC], lw=2, label=["marginal cost" "average cost"], xlabel="q", ylims=(0, 30))
```
