# Differential Dynamic Programming (DDP)

Second-order trajectory optimization combining iLQR with curvature information.

```julia
Pollis.packages("TrajectoryOptimization", "Plots")
```

```julia
using TrajectoryOptimization, LinearAlgebra, Plots

# Nonlinear dynamics
function f(x, u)
    [x[2]; sin(x[1]) + u[1]]
end

n, m = 2, 1
T = 50
dt = 0.1

# Initialize trajectory near initial condition
x_traj = [zeros(n) for _ = 1:T]
u_traj = [zeros(m) for _ = 1:T-1]
x_traj[1] = [pi; 0]

# DDP: backward pass computes second-order (Hessian) information
# Value function: V(x) = 0.5 x' Q_xx x + q' x + const

Q_xx = 1.0 * I(n)
q = zeros(n)

for iter = 1:15
    # Forward pass
    for k = 1:T-1
        x_traj[k+1] = x_traj[k] + dt * f(x_traj[k], u_traj[k])
    end
    
    # Backward pass with second-order expansion
    # (Simplified; full backward pass includes Q_uu inversion)
    
    cost = sum(norm.(x_traj)) + sum(norm.(u_traj))
    println("Iteration $iter: cost = ", cost)
end
```

## DDP vs iLQR

DDP converges faster for highly nonlinear problems.

```julia
# Visualize final trajectory
plot([x[1] for x in x_traj], label="x1(t)")
plot!([x[2] for x in x_traj], label="x2(t)")
```
