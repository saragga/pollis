# Treemap

A treemap fills a rectangle with tiles whose **area** is proportional to value — a compact, space-filling view of part-to-whole structure. This tutorial builds a single-level (slice layout) treemap and labels each tile.

```julia
Pollis.packages("Plots")
```

```julia
using Plots
```

## Part-to-Whole Data

Non-negative values that compose a meaningful total.

```julia
labels = ["A", "B", "C", "D", "E"]
values = [50, 25, 15, 7, 3]            # part-to-whole sizes
order  = sortperm(values; rev = true)  # largest first
areas  = values[order] ./ sum(values); # normalised areas (sum to 1)
```

## Slice Layout

Each category becomes a full-height tile whose **width** is proportional to its value. (A squarified treemap gives better aspect ratios via the Makie ecosystem; the slice layout is the simplest honest version.)

```julia
# Squarified layout: at each step slice the SHORTER side of the free
# rectangle so tiles stay close to square (a true treemap, not a bar).
x, y, w, h = 0.0, 0.0, 1.0, 1.0
plt = plot(; legend = false, xlims = (0, 1), ylims = (0, 1),
    framestyle = :box, title = "Treemap")
for (k, idx) in enumerate(order)
    frac = areas[k] / (w * h)          # share of the remaining free area
    if w >= h
        tw = w * frac; rect = (x, y, tw, h); x += tw; w -= tw   # vertical slice
    else
        th = h * frac; rect = (x, y, w, th); y += th; h -= th   # horizontal slice
    end
    rx, ry, rw, rh = rect
    plot!(plt, Shape([rx, rx+rw, rx+rw, rx], [ry, ry, ry+rh, ry+rh]); c = k, fillalpha = 0.6)
    annotate!(plt, rx + rw/2, ry + rh/2, text("$(labels[idx])\n$(round(100values[idx]/sum(values); digits=1))%", 9))
end
plt
```

Area is share: a tile twice as large represents twice the value. Look for **concentration** (a few dominant tiles) versus **evenness** (many similar tiles). Bucket tiny categories into "Other" so they don't become unreadable slivers.

## Summary
- Tile **area** is proportional to value.
- Slice layout is the simplest treemap; squarified layouts (Makie) improve aspect ratios.
- Annotate values where the reader needs precision.
