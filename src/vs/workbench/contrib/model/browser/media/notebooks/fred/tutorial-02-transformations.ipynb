# Units and Frequency

FRED can transform a series **server-side** before it reaches Julia. Pass `units` and `frequency` to `get_data` instead of computing growth rates or resampling by hand.

```julia
Pollis.packages("FredData")
```

```julia
# ENV["FRED_API_KEY"] = "YOURKEY"  # uncomment and set if not already in the environment
using FredData

f = Fred()
```

## Levels vs Year-over-Year

The Consumer Price Index (`CPIAUCSL`) is an index level. Request `units="pc1"` for year-over-year inflation.

```julia
cpi_level = get_data(f, "CPIAUCSL")
cpi_yoy   = get_data(f, "CPIAUCSL"; units="pc1")

println(last(cpi_level.data, 3))
println(last(cpi_yoy.data, 3))
```

## Other Units

| `units` | Meaning |
|---|---|
| `lin` | Levels (default) |
| `chg` | Change from previous period |
| `pch` | Percent change |
| `pc1` | Percent change from a year ago |
| `pca` | Compounded annual rate of change |

```julia
# Real GDP as a compounded annual growth rate
gdp_growth = get_data(f, "GDPC1"; units="pca")
last(gdp_growth.data, 5)
```

## Resampling Frequency

`frequency` resamples to a coarser frequency; `aggregation_method` (`avg`, `sum`, `eop`) says how.

```julia
# Daily fed funds rate (DFF), averaged into a monthly series
funds = get_data(f, "DFF"; frequency="m", aggregation_method="avg")
last(funds.data, 5)
```

## Summary
- `units` applies a transform: `pc1` (year-over-year), `pca` (annualised growth), `chg`, `log`, …
- `frequency` + `aggregation_method` resample to a coarser frequency.
- Transforms run on FRED's servers, so the returned `units` field reflects what you asked for.
