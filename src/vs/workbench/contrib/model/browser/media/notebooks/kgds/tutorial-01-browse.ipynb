# Browse the Kaggle Catalogue

Search, sort, and filter Kaggle datasets over the public [REST API](https://www.kaggle.com/docs/api) — pure Julia with `HTTP.jl` and `JSON3.jl`, no credentials needed. A token is only required later for downloads.

```julia
Pollis.packages("HTTP", "JSON3")
```

```julia
# import Pkg; Pkg.add(["HTTP", "JSON3"])
using HTTP, JSON3
get_json(url) = JSON3.read(HTTP.get(url).body)
```

## Most-Voted Datasets

`sortBy` is one of `hottest`, `votes`, `updated`, or `published`.

```julia
res = get_json("https://www.kaggle.com/api/v1/datasets/list?sortBy=votes&pageSize=10")
for d in res
    println(d.ref, "  votes=", d.voteCount, "  downloads=", d.downloadCount)
end
```

## Search and Filter

Use `search` for free text and `fileType` / `license` to narrow the results.

```julia
q = HTTP.escapeuri("housing prices")
url = "https://www.kaggle.com/api/v1/datasets/list?search=$q&fileType=csv&license=cc&sortBy=votes&pageSize=10"
for d in get_json(url)
    println(d.ref, "  ", d.licenseName, "  ", d.totalBytes, " bytes")
end
```

## Read One Dataset's Metadata

The list endpoint already returns rich metadata; search by slug to inspect a single dataset.

```julia
d = first(get_json("https://www.kaggle.com/api/v1/datasets/list?search=iris&pageSize=5"))
println("ref:       ", d.ref)
println("title:     ", d.title)
println("license:   ", d.licenseName)
println("downloads: ", d.downloadCount, ",  votes: ", d.voteCount)
```

## Summary
- `/datasets/list` is public — `sortBy`, `search`, `fileType`, and `license` shape the results.
- Each dataset carries `ref`, `title`, `licenseName`, `totalBytes`, `voteCount`, and `downloadCount`.
- A token is only needed to list files and download (see the next tutorial).
