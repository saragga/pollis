# Bubble Chart

A bubble chart encodes a third variable as marker size and a fourth as colour, packing four dimensions into one scatter.

```julia
using Plots
using Random

Random.seed!(22);
```

## Four Variables

Position (x, y), size, and colour. **Map the size variable to area** — radius proportional to its square root — so the eye reads magnitude correctly.

```julia
n = 60
x = randn(n)
y = randn(n)
sz = rand(n) .* 100      # third variable
col = x .+ y             # fourth variable

radii = 12 .* sqrt.(sz ./ maximum(sz))
scatter(x, y; markersize = radii, zcolor = col, alpha = 0.6,
    xlabel = "X", ylabel = "Y", title = "Bubble Chart", label = "", colorbar_title = "x + y")
```

Read position first, then check whether bubble size grows in any direction and whether colours cluster.

## Why Area, Not Radius

If size mapped to radius linearly, a value twice as large would draw a bubble **four times** the area — visually overstating it. Square-rooting fixes this.

```julia
# Wrong (radius linear in value) vs right (area linear in value):
bad  = 12 .* (sz ./ maximum(sz))
good = 12 .* sqrt.(sz ./ maximum(sz))
p1 = scatter(x, y; markersize = bad,  title = "radius ~ value (overstates)", label = "", alpha = 0.5)
p2 = scatter(x, y; markersize = good, title = "area ~ value (correct)",      label = "", alpha = 0.5)
plot(p1, p2; layout = (1, 2), size = (900, 360))
```

## Summary
- Four channels: x, y, size, colour — near the legibility limit.
- Map value to area (radius ~ sqrt(value)).
- Keep the size range moderate so all bubbles stay readable.
