# Forex and Crypto

Fetch foreign-exchange rates and digital-currency prices with [AlphaVantage.jl](https://github.com/ellisvalentiner/AlphaVantage.jl).

Set `ENV["ALPHA_VANTAGE_API_KEY"]` for this session first (see the Authentication wiki).

```julia
Pollis.packages("AlphaVantage")
```

```julia
using AlphaVantage
```

## Exchange Rate and FX Series

```julia
rate = currency_exchange_rate("USD", "EUR")
println(rate)
```

```julia
fx = fx_daily("EUR", "USD"; outputsize="compact")
println(fx)
```

## Digital Currency

```julia
btc = digital_currency_daily("BTC", "USD")
println(btc)
```

## Summary
- `currency_exchange_rate(from, to)` gives a real-time rate.
- `fx_daily(from, to)` returns a daily FX series.
- `digital_currency_daily(sym, market)` returns daily crypto prices.
