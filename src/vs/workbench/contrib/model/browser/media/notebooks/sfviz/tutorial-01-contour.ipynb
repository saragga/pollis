# Contour Plot

A contour plot draws the level curves (isolines) of a 2D scalar field `f(x, y)` — every point on a line shares the same value, like contour lines on a topographic map. 2D contours render statically with [CairoMakie](https://github.com/MakieOrg/Makie.jl), so they show up inline.

```julia
using CairoMakie
```

## A Scalar Field

Build `zs[i, j] = f(xs[i], ys[j])` on a grid.

```julia
xs = LinRange(0, 10, 100)
ys = LinRange(0, 15, 100)
zs = [cos(x) * sin(y) for x in xs, y in ys];
```

## Draw the Contours

Pass explicit `levels` to control which values get a line; `labels = true` annotates them.

```julia
contour(xs, ys, zs; levels = -1:0.1:1, labels = true, colormap = :viridis)
```

**Line spacing is slope**: lines close together mean a steep gradient, far apart means flat. Closed loops enclose a local maximum or minimum.

## Custom Levels and Colour Scale

For a field that spans orders of magnitude, use logarithmic levels and a reversible colour scale.

```julia
himmelblau(x, y) = (x^2 + y - 11)^2 + (x + y^2 - 7)^2
u = v = range(-6, 6; length = 100)
w = himmelblau.(u, v')
levels = 10.0 .^ range(0.3, 3.5; length = 10)
contour(u, v, w; labels = true, levels, colormap = :hsv)
```

## Filled Contours

`contourf` shades the bands between levels — easier for overall shape. Overlay `contour!` to keep the crisp lines too.

```julia
fig, ax, _ = contourf(xs, ys, zs; levels = 12, colormap = :viridis)
contour!(ax, xs, ys, zs; levels = 12, color = :white, linewidth = 0.5)
fig
```

## Summary
- Contour lines are levels of a 2D field; spacing encodes the gradient.
- Choose `levels` deliberately (linear, log, custom).
- `contourf` fills the bands; CairoMakie renders 2D contours inline.
