# Browse the Hugging Face Hub

Search the dataset catalogue and read a dataset's metadata over the public [Hub REST API](https://huggingface.co/docs/hub/api) — pure Julia with `HTTP.jl` and `JSON3.jl`, no Python. Public datasets need no token; click **Set token** for gated or private ones.

```julia
Pollis.packages("HTTP", "JSON3")
```

```julia
# import Pkg; Pkg.add(["HTTP", "JSON3"])
using HTTP, JSON3

auth = haskey(ENV, "HUGGING_FACE_HUB_TOKEN") ? ["Authorization" => "Bearer $(ENV["HUGGING_FACE_HUB_TOKEN"])"] : Pair{String, String}[]
get_json(url) = JSON3.read(HTTP.get(url, auth).body)
```

## List the Most-Downloaded Datasets

Sort by `downloads` descending and ask for full metadata on each item.

```julia
url = "https://huggingface.co/api/datasets?sort=downloads&direction=-1&limit=10&full=true"
for d in get_json(url)
    println(d.id, "  downloads=", get(d, :downloads, 0))
end
```

## Search and Filter

Use `search` for free text, or `filter=<group>:<value>` for tags such as task category.

```julia
q = HTTP.escapeuri("sentiment")
for d in get_json("https://huggingface.co/api/datasets?search=$q&limit=10")
    println(d.id)
end
```

## Read One Dataset's Metadata

The `/api/datasets/<id>` endpoint returns tags, licence, downloads, and more.

```julia
info = get_json("https://huggingface.co/api/datasets/imdb")
println("id:        ", info.id)
println("downloads: ", get(info, :downloads, 0))
println("tags:      ", get(info, :tags, String[]))
```

## Summary
- `/api/datasets` lists and searches the catalogue; `sort`, `direction`, `limit`, and `filter` shape the results.
- `/api/datasets/<id>` returns full metadata for one dataset.
- Add an `Authorization: Bearer` header only when a token is set.
