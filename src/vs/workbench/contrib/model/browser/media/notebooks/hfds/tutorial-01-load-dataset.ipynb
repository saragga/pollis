# Load a Hugging Face Dataset

Load a dataset split into Julia with [HuggingFaceDatasets.jl](https://github.com/CarloLucibello/HuggingFaceDatasets.jl), convert it to native Julia objects, and inspect the rows. The package wraps the Python `datasets` library through PythonCall, so the first call provisions a one-time Conda environment.

```julia
# import Pkg; Pkg.add("HuggingFaceDatasets")
using HuggingFaceDatasets
```

## Load a Split

`load_dataset` takes a dataset id and a `split`. Append `.with_format("julia")` to get native Julia objects instead of Python ones. For gated datasets, set a token first (see the **Authentication** wiki).

```julia
train = load_dataset("imdb", split = "train").with_format("julia")
println("rows: ", length(train))
```

## Inspect Observations

A `Dataset` behaves like a Julia collection: index it, slice it, and read field names with `keys`.

```julia
sample = train[1]
println("fields: ", keys(sample))

for i in 1:3
    println(i, "  label=", train[i]["label"])
end
```

## All Splits at Once

Omit `split` to load a `DatasetDict` keyed by split name.

```julia
dd = load_dataset("imdb")
println(dd)

test = dd["test"].with_format("julia")
println("test rows: ", length(test))
```

## Summary
- `load_dataset(id; split)` loads one split; omit `split` for a `DatasetDict`.
- `.with_format("julia")` returns native Julia objects.
- A `Dataset` supports `length`, indexing, slicing, and `keys` on each observation.
