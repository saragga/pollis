# QQ Plots

A quantile-quantile plot is the most informative normality check: points on the 45-degree line mean the sample matches the reference; systematic departures reveal skew and heavy tails.

```julia
using StatsPlots
using Distributions
using Random

Random.seed!(11);
```

## Normal Data

`qqnorm` compares a sample to the Normal distribution.

```julia
x = randn(300)
qqnorm(x; xlabel = "Theoretical quantiles", ylabel = "Sample quantiles", title = "Normal sample")
```

## Heavy Tails

A t-distribution with few degrees of freedom is heavier-tailed than normal — the QQ plot shows the characteristic S-shape, ends peeling away from the line.

```julia
xt = rand(TDist(3), 300)
qqnorm(xt; title = "Heavy-tailed (t, df=3)", xlabel = "Theoretical quantiles", ylabel = "Sample quantiles")
```

## Skew

A log-normal sample is right-skewed — the QQ plot curves away from the line.

```julia
xs = rand(LogNormal(0, 0.6), 300)
qqnorm(xs; title = "Right-skewed (log-normal)", xlabel = "Theoretical quantiles", ylabel = "Sample quantiles")
```

## Comparing Against Any Distribution

`qqplot(D, x)` compares against an arbitrary reference, here an Exponential.

```julia
xe = rand(Exponential(1.0), 300)
qqplot(Exponential, xe; title = "Exponential reference", xlabel = "Theoretical quantiles", ylabel = "Sample quantiles")
```

## Summary
- Points on the line → match.
- S-shape → heavier/lighter tails.
- Curvature → skew.
- Pair the plot with a formal test (KS, Anderson-Darling).
