# Nightingale Rose

A nightingale rose (polar area / coxcomb chart) is a bar chart wrapped onto polar axes: equal-angle sectors whose **radius** encodes magnitude. Florence Nightingale used it to show causes of mortality by month. It is best for **cyclic** categories.

```julia
Pollis.packages("Plots")
```

```julia
using Plots
```

## Cyclic Data

Twelve months of a magnitude (e.g. incidents per month). The categories wrap around a year, which is why a circular layout fits.

```julia
months = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
values = [12, 19, 9, 25, 17, 22, 30, 28, 15, 11, 8, 14]
```

## The Rose Chart

Drawing a bar series in polar projection produces the rose: each sector spans an equal angle and extends to its value.

```julia
plt = plot(; aspect_ratio = :equal, legend = false, framestyle = :none,
    title = "Nightingale Rose")
n = length(values)
for i in 1:n
    a0, a1 = 2pi * (i - 1) / n, 2pi * i / n     # equal angular slice
    ts = range(a0, a1; length = 24)             # sample the arc
    xs = [0.0; values[i] .* cos.(ts); 0.0]      # wedge: centre -> arc -> centre
    ys = [0.0; values[i] .* sin.(ts); 0.0]      # radius encodes the value
    plot!(plt, Shape(xs, ys); c = i, fillalpha = 0.7, linecolor = :white)
    amid = (a0 + a1) / 2
    annotate!(plt, (values[i] + 2) * cos(amid), (values[i] + 2) * sin(amid),
              text(months[i], 7))
end
plt
```

## A Caution on Area

The eye reads **area**, and a sector's area grows with the *square* of its radius. A month with twice the value looks far more than twice as large. The rose is excellent for spotting a *cyclic pattern*; for precise magnitude comparison a plain bar chart is more honest.

```julia
bar(months, values; legend = false, xrotation = 45,
    ylabel = "Value", title = "Same data, linear bars")
```

## Summary
- Equal-angle sectors; **radius** encodes magnitude.
- Designed for cyclic categories (months, hours, directions).
- Area exaggerates large values — use for pattern, not precise comparison.
