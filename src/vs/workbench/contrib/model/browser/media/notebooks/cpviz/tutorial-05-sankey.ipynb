# Sankey Diagram

A Sankey diagram shows quantities flowing between nodes as bands whose **width is proportional to the flow**. Quantity is conserved as it moves from sources through intermediate nodes to sinks. This tutorial uses [SankeyMakie.jl](https://github.com/MakieOrg/SankeyMakie.jl).

```julia
Pollis.packages("SankeyMakie", "CairoMakie")
```

```julia
using SankeyMakie
using CairoMakie
```

## A Flow Network

Edges are `(source, target, weight)` with 1-based node indices. Here two energy sources feed a grid, which then splits between homes and industry.

```julia
connections = [(1, 3, 8.0),   # Coal  -> Grid
               (2, 3, 4.0),   # Gas   -> Grid
               (3, 4, 7.0),   # Grid  -> Homes
               (3, 5, 5.0)]   # Grid  -> Industry
names = ["Coal", "Gas", "Grid", "Homes", "Industry"];
```

## Drawing the Sankey

Band width is proportional to flow. At a conserving node, total inflow equals total outflow.

```julia
sankey(connections; nodelabels = names)
```

## Checking Conservation

The grid (node 3) receives 8 + 4 = 12 and sends 7 + 5 = 12. When inflow and outflow disagree at a node, that usually signals a data error rather than a real imbalance.

```julia
inflow  = sum(w for (s, t, w) in connections if t == 3)
outflow = sum(w for (s, t, w) in connections if s == 3)
println("Grid inflow = ", inflow, "  outflow = ", outflow)
```

## Summary
- Band **width** encodes flow magnitude.
- Follow bands to see how flows split and merge.
- Inflow should equal outflow at each node; reorder nodes to reduce crossings.
