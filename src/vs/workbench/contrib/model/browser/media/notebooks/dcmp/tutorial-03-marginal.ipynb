# Marginal Plots

A marginal plot combines a joint scatter of two variables with their marginal distributions on the axes — the relationship and both one-dimensional shapes in one figure.

```julia
Pollis.packages("StatsPlots")
```

```julia
using StatsPlots
using Random

Random.seed!(12);
```

## Marginal Histogram

`marginalhist` draws the joint scatter with histograms on each axis.

```julia
x = randn(500)
y = 0.7 .* x .+ 0.7 .* randn(500)

marginalhist(x, y; bins = 30, xlabel = "X", ylabel = "Y")
```

The centre shows a positive linear relationship; the margins confirm both variables are roughly normal.

## Marginal KDE

`marginalkde` uses smooth kernel density margins instead of histograms — better when you care about shape and have enough data.

```julia
marginalkde(x, y; xlabel = "X", ylabel = "Y")
```

## Skewed Margins

When a variable is skewed, the margin makes it obvious — a transformation candidate before regression.

```julia
ys = exp.(0.5 .* y)   # skew Y
marginalhist(x, ys; bins = 30, xlabel = "X", ylabel = "exp(Y)")
```

## Summary
- Centre → joint relationship; margins → each variable's distribution.
- Histogram margins for counts, KDE margins for smooth shape.
- Skewed margins flag transformation candidates.
