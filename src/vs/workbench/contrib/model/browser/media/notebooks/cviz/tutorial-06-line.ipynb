# Line Plot

A line plot connects points in order with straight segments. In the statistical setting its most common job is to draw a **fitted line** — a regression fit or trend — over the raw observations. Uses StatsPlots.jl.

```julia
Pollis.packages("StatsPlots")
```

```julia
using StatsPlots, Statistics, Random
```

## Noisy Linear Data

A linear signal `y = 2x + 1` plus Gaussian noise. Sort `x` so the line can be drawn cleanly left to right.

```julia
Random.seed!(42)
x = sort(randn(80))
y = 2 .* x .+ 1 .+ randn(80);
```

## Fit a Line by Least Squares

The OLS slope is `cov(x, y) / var(x)` and the intercept passes the line through the means.

```julia
b1 = cov(x, y) / var(x)
b0 = mean(y) - b1 * mean(x)
(b0, b1)
```

## Observations and Fit

Scatter the data, then connect the fitted values `b0 .+ b1 .* x` with a line.

```julia
scatter(x, y; label = "Observations", alpha = 0.5)
plot!(x, b0 .+ b1 .* x; label = "Fitted line", lw = 2)
```

The slope is the rate of change of `y` with `x`; the intercept is `y` at `x = 0`. The points should scatter evenly above and below the line — a systematic curve away from it means a straight fit is too simple.

## Summary
- A line plot connects ordered points; sort by `x` first.
- OLS: `b1 = cov(x, y)/var(x)`, `b0 = mean(y) - b1*mean(x)`.
- Overlay `plot!` on a `scatter` to show a fitted regression line; check residuals before trusting it.
