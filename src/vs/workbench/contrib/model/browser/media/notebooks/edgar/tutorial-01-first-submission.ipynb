# Your First EDGAR Submission

List a company's recent filings from the SEC [EDGAR APIs](https://www.sec.gov/search-filings/edgar-application-programming-interfaces). The `data.sec.gov` APIs return JSON and need **no API key** — but every request must carry a `User-Agent` header with your name and a contact email, or the SEC returns HTTP 403.

```julia
Pollis.packages("EDGAR" => "https://github.com/Trumpingtons/EDGAR.jl")
```

## Resolve a Ticker to a CIK

Filers are keyed by **CIK**. `EDGAR.jl`'s `cik_for_ticker` looks a ticker up in the SEC's `company_tickers.json` map and returns the 10-digit, zero-padded CIK.

```julia
cik = cik_for_ticker("AAPL")     # "0000320193" — data.sec.gov wants 10 digits
```

## Fetch the Submission History

The `/submissions/` endpoint returns the filer's identity plus a `filings.recent` block of **parallel arrays** (one entry per filing).

```julia
sub = fetch_submissions(cik)

println(sub.name, " - ", sub.sicDescription)
println("Tickers: ", sub.tickers)
```

## List Recent Filings

`filings.recent` has aligned vectors: `form`, `filingDate`, `accessionNumber`, `primaryDocument`. Index them together.

```julia
recent = sub.filings.recent
for i in 1:8
    println(recent.filingDate[i], "  ", rpad(recent.form[i], 8), recent.primaryDocument[i])
end
```

## Summary
- The `data.sec.gov` APIs need no key, but a `User-Agent` header is mandatory.
- Resolve a ticker to a 10-digit CIK with `company_tickers.json`.
- `/submissions/CIK##########.json` returns `filings.recent` as parallel arrays of form, date, accession, and document.
