# Pie Chart

A pie chart shows how a few **categories** make up a whole, each slice's angle proportional to its share. Best for a quick part-to-whole read with few slices; for precise comparison a bar chart is usually clearer. Uses [CairoMakie](https://github.com/MakieOrg/Makie.jl).

```julia
using CairoMakie
```

## Parts of a Whole

Each value is one slice; the colours label them.

```julia
data   = [36, 12, 68, 5, 42, 27]
colors = [:yellow, :orange, :red, :blue, :purple, :green];
```

## A Donut Pie

`inner_radius > 0` opens the centre into a donut, which many find easier to compare. `strokecolor`/`strokewidth` draw white gaps between slices, and `autolimitaspect = 1` keeps the slices circular.

```julia
f, ax, plt = pie(data; color = colors, radius = 4, inner_radius = 2,
                 strokecolor = :white, strokewidth = 5,
                 axis = (autolimitaspect = 1,))
Legend(f[1, 2], ax)
f
```

The angle of each slice is its share of the total. Keep to a handful of slices — people judge angle and area poorly, so for many categories or precise values reach for a bar chart instead.

## Summary
- A pie chart maps each category's share to a slice angle.
- `inner_radius > 0` makes a donut; `Legend(f[1, 2], ax)` adds a key.
- Use few slices; prefer a bar chart for precise comparison.
