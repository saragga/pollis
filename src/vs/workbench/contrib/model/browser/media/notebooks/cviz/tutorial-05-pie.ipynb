# Pie Chart

A pie chart shows how a few **categories** make up a whole, each slice's angle proportional to its share. Best for a quick part-to-whole read with few slices; for precise comparison a bar chart is usually clearer. Uses [StatsPlots](https://github.com/JuliaPlots/StatsPlots.jl), which includes all of [Plots](https://github.com/JuliaPlots/Plots.jl).

```julia
Pollis.packages("StatsPlots")
```

```julia
using StatsPlots
```

## Parts of a Whole

Each value is one slice; the labels name them and the colours tell them apart.

```julia
labels = ["Yellow", "Orange", "Red", "Blue", "Purple", "Green"]
data   = [36, 12, 68, 5, 42, 27]
colors = [:yellow, :orange, :red, :blue, :purple, :green];
```

## Drawing the Pie

`pie(labels, data)` draws one slice per value and a legend with the labels. `linecolor`/`linewidth` draw white gaps between slices, and `aspect_ratio = :equal` keeps the slices circular.

```julia
pie(labels, data;
    color = colors, linecolor = :white, linewidth = 3,
    aspect_ratio = :equal, legend = :outerright)
```

The angle of each slice is its share of the total. Keep to a handful of slices — people judge angle and area poorly, so for many categories or precise values reach for a bar chart instead.

## Summary
- A pie chart maps each category's share to a slice angle.
- `pie(labels, data)` draws the slices and a legend; `legend = :outerright` moves the legend beside the pie.
- Use few slices; prefer a bar chart for precise comparison.
