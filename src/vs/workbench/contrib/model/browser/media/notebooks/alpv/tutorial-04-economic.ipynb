# Economic Indicators

Pull US macroeconomic series from Alpha Vantage with [AlphaVantage.jl](https://github.com/ellisvalentiner/AlphaVantage.jl).

Set `ENV["ALPHA_VANTAGE_API_KEY"]` for this session first.

```julia
Pollis.packages("AlphaVantage")
```

```julia
using AlphaVantage
```

## GDP and CPI

```julia
gdp = real_gdp(; interval="annual")
inflation = cpi(; interval="monthly")
println(gdp)
println(inflation)
```

## Treasury Yields and Policy Rate

```julia
yield10y = treasury_yield(; interval="monthly", maturity="10year")
funds    = federal_fund_rate(; interval="monthly")
println(yield10y)
println(funds)
```

## Summary
- `real_gdp`, `cpi`, `treasury_yield`, and `federal_fund_rate` return dated macro series.
- `interval` selects frequency (`"annual"`, `"quarterly"`, `"monthly"`).
- `treasury_yield` accepts a `maturity` such as `"10year"`.
