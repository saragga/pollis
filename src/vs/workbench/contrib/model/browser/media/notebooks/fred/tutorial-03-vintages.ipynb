# Vintages and Revisions

Many macro series are **revised** after first publication. FRED keeps every past **vintage**; ALFRED ("Archival FRED") lets you fetch the data *as it looked* on a given date with the `vintage_dates` keyword.

This matters for backtesting: a model should only see the numbers that were actually available at decision time.

```julia
# ENV["FRED_API_KEY"] = "YOURKEY"  # uncomment and set if not already in the environment
using FredData

f = Fred()
```

## Today's Values

By default `get_data` returns the latest, fully revised data. Real GDP is a heavily revised series.

```julia
latest = get_data(f, "GDPC1")
last(latest.data, 4)
```

## A Single Vintage

Pass `vintage_dates` to ask: *what did this series look like on that date?* The `realtime_start` / `realtime_end` columns record the vintage window.

```julia
# Real GDP as it was published on 2014-06-01
vintage = get_data(f, "GDPC1"; vintage_dates="2014-06-01")
last(vintage.data, 4)
```

## Comparing Two Vintages

Request several vintage dates at once to see how an estimate was revised over time.

```julia
revisions = get_data(f, "GDPC1"; vintage_dates="2014-06-01,2015-06-01,2020-06-01")
first(revisions.data, 8)
```

## Summary
- Without `vintage_dates`, you get today's revised values.
- `vintage_dates="YYYY-MM-DD"` returns the data as published on that date (ALFRED).
- Pass a comma-separated list to compare revisions; `realtime_start` / `realtime_end` mark each vintage.
