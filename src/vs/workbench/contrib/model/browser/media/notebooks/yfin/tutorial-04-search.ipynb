# Symbol and News Search

Resolve ticker symbols and scan recent news with [YFinance.jl](https://github.com/eohne/YFinance.jl).

```julia
Pollis.packages("YFinance")
```

```julia
using YFinance
```

## Resolve a Symbol

`get_symbols` takes a company or security name and returns candidate tickers.

```julia
hits = get_symbols("micro")
println(hits)
```

## List an Exchange

`get_all_symbols` returns every symbol listed on a given exchange.

```julia
nyse = get_all_symbols("NYSE")
println("NYSE symbols: ", length(nyse))
```

## Latest News

`search_news` returns a news collection; read it with `titles`, `links`, and `timestamps`.

```julia
news = search_news("MSFT")
for (title, link) in zip(titles(news), links(news))
    println(title)
    println("  ", link)
end
```

## Summary
- `get_symbols(name)` resolves a name to candidate tickers.
- `get_all_symbols(exchange)` lists an entire exchange.
- `search_news(query)` returns headlines; use `titles`, `links`, and `timestamps`.
