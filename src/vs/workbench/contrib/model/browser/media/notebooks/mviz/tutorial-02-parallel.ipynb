# Parallel Coordinates

Each variable is a vertical axis and each observation a line crossing all axes. Colouring by group exposes clusters in high-dimensional data.

```julia
Pollis.packages("Plots")
```

```julia
using Plots
using Random
using Statistics

Random.seed!(21);
```

## Two-Group Dataset

Four variables, two groups with different means. We standardise each variable so no axis dominates.

```julia
n = 60
g1 = hcat(randn(n).+2, randn(n), randn(n).-1, randn(n).+1)
g2 = hcat(randn(n).-2, randn(n).+1, randn(n).+1, randn(n).-1)
X = vcat(g1, g2)
grp = vcat(fill(1, n), fill(2, n))

# standardise columns
Z = (X .- mean(X, dims = 1)) ./ std(X, dims = 1);
```

## Plot the Lines

Each row is a polyline across the four axes, coloured by group.

```julia
plt = plot(; xlabel = "Variable", ylabel = "Standardised value", title = "Parallel Coordinates", legend = :topright)
for i in 1:size(Z, 1)
    plot!(plt, 1:4, Z[i, :]; color = grp[i], alpha = 0.3, label = "")
end
plt
```

The two groups form distinct bundles, separating most clearly on variable 1. Crossing patterns between adjacent axes indicate negative correlation; parallel runs indicate positive correlation.

## Summary
- Standardise first so each axis is comparable.
- Bundles = clusters; X-crossings = negative correlation.
- Only adjacent axes compare directly — reorder to probe other pairs.
