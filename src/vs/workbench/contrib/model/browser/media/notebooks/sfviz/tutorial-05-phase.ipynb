# Phase Portrait

A phase portrait shows the flow of a 2D **dynamical system** `dx/dt = f(x)`. `streamplot` integrates trajectories through the vector field, revealing fixed points, spirals, saddles, and limit cycles. Uses [CairoMakie](https://github.com/MakieOrg/Makie.jl) (swap in `GLMakie` for an interactive window).

```julia
Pollis.packages("CairoMakie")
```

```julia
using CairoMakie
```

## The Vector Field

The state is `x = (theta, omega)`; `field` returns the velocity `dx/dt` as a `Point2f`. This is a damped pendulum.

```julia
field(x) = Point2f(x[2], -sin(x[1]) - 0.2 * x[2])
```

## Streamplot

`streamplot` seeds trajectories across the domain and integrates them through the field. Colour encodes the local speed `|f(x)|`; `gridsize` sets the seed density.

```julia
streamplot(field, -2pi .. 2pi, -4 .. 4; colormap = :plasma, gridsize = (32, 32))
```

Follow the streamlines: trajectories spiral into the damped equilibria at `theta = 0, 2pi, ...` and swirl around the saddles in between. The arrows carry the direction of time, which a static image otherwise hides.

## Summary
- `field(x)` must return a `Point2f` velocity, not a scalar.
- `streamplot(field, xrange, yrange)` draws the flow; colour = speed.
- Pick the domain to bracket the fixed points you care about.
