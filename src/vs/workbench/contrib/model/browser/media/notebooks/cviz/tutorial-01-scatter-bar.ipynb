# Scatter and Bar

Scatter plots show the relationship between two numeric variables; bar charts compare a value across categories. This tutorial uses [StatsPlots.jl](https://github.com/JuliaPlots/StatsPlots.jl) and [Plots.jl](https://github.com/JuliaPlots/Plots.jl).

```julia
Pollis.packages("Plots", "StatsPlots")
```

```julia
using Plots
using StatsPlots
using Random
using Statistics

Random.seed!(1);
```

## Scatter Plot

One point per observation. We add a third categorical variable through colour to expose group structure.

```julia
n = 200
x = randn(n)
y = 1.5 .* x .+ randn(n)
group = rand(["A", "B"], n)

scatter(x, y; group = group, xlabel = "X", ylabel = "Y",
    title = "Scatter by Group", alpha = 0.6, legend = :topleft)
```

Direction (upward), form (linear), and strength (a fairly tight band) are all read straight off the cloud. Colouring by group checks whether the relationship differs between sub-populations.

## Bar Chart

A bar chart compares a single value per category. Here we aggregate the mean of `y` within each group.

```julia
cats = sort(unique(group))
means = [mean(y[group .== c]) for c in cats]

bar(cats, means; xlabel = "Group", ylabel = "Mean of Y",
    title = "Group Means", legend = false, ylims = (minimum(means) < 0 ? :auto : (0, Inf)))
```

**Always anchor a bar axis at zero** for counts and magnitudes so differences are not visually exaggerated.

## Summary
- Scatter → relationship between two numeric variables.
- Bar → magnitude comparison across categories.
- Encode a categorical third variable with colour.
