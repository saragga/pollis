# Prices and Charts

Download OHLC price bars from Yahoo Finance with [YFinance.jl](https://github.com/eohne/YFinance.jl), sink them into a [TimeArray](https://github.com/JuliaStats/TimeSeries.jl), and plot the adjusted close.

```julia
using YFinance, TimeSeries, Tables
```

## Download Daily Bars

`get_prices` returns an `OrderedDict` of column vectors. Choose a `range` and an `interval` (see the Constraints wiki for valid values).

```julia
prices = get_prices("AAPL"; range="1mo", interval="1d")
println("Ticker:     ", prices["ticker"])
println("Last close: ", last(prices["close"]))
```

## Sink into a TimeArray

Pass `TimeArray` as the first argument to get a time-indexed result. A `TimeArray` implements the Tables.jl interface.

```julia
ta = get_prices(TimeArray, "AAPL"; range="6mo", interval="1d")
for row in first(Tables.rows(ta), 3)
    println(row.timestamp, "  close=", row.close)
end
```

## Plot the Adjusted Close

```julia
using Plots
plot(ta[:adjclose]; label="AAPL", lw=2, legend=:topleft)
title!("Adjusted Close")
```

## Summary
- `get_prices(symbol; range, interval)` returns an `OrderedDict` of OHLC column vectors.
- Pass `TimeArray` first to sink directly into a time-indexed table.
- Use `adjclose` for returns; it is corrected for dividends and splits.
