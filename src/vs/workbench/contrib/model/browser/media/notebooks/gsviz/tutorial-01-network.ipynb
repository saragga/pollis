# Network Diagram

A network diagram draws a graph: nodes (entities) joined by edges (relationships), placed by a layout so connected nodes sit close together. This tutorial uses [Graphs.jl](https://github.com/JuliaGraphs/Graphs.jl) for the graph and [GraphMakie.jl](https://github.com/MakieOrg/GraphMakie.jl) to draw it.

```julia
using Graphs
using GraphMakie
using CairoMakie
using Random

Random.seed!(42);   # node positions are arbitrary; seed for a reproducible layout
```

## Build a Graph

A random Erdos-Renyi graph: 14 nodes, each possible edge present with probability 0.18. In practice you would build edges from your data with `add_edge!`.

```julia
g = erdos_renyi(14, 0.18)
deg = degree(g)          # number of connections per node
```

## Draw It

`graphplot` uses a force-directed layout by default. We size and colour nodes by degree so hubs stand out.

```julia
graphplot(g;
    node_size  = 8 .+ 2 .* deg,
    node_color = deg,
    nlabels    = string.(1:nv(g)))
```

Read **connectivity first, position second** — proximity suggests relatedness, but the absolute coordinates carry no meaning. Hubs (large nodes) have many connections; densely connected groups form visible clusters.

## Directed Edges

For dependencies or flows, use a directed graph and show arrowheads.

```julia
dg = SimpleDiGraph(5)
add_edge!.(Ref(dg), [1, 1, 2, 3, 4], [2, 3, 4, 4, 5])   # source -> target
graphplot(dg; arrow_show = true, nlabels = string.(1:5))
```

## Summary
- Nodes = entities, edges = relationships; the layout places connected nodes nearby.
- Size/colour nodes by a metric (degree, centrality) to surface hubs.
- Positions are arbitrary — fix the seed for reproducible figures.
- Use a `SimpleDiGraph` + `arrow_show` for directed relationships.
