# Conditional Formatting

This tutorial applies rule-based cell highlighting using `XLSX.setConditionalFormat` (XLSX.jl v0.11+). Rules are stored in the file and rendered by Excel — cells change colour automatically when the workbook is opened.

```julia
Pollis.packages("XLSX")
```

```julia
using XLSX
```

## Write a Sales Table

```julia
sales = (
    product = ["Alpha", "Beta", "Gamma", "Delta", "Epsilon"],
    sales   = [5200.0, 800.0, 3100.0, 450.0, 7800.0],
    margin  = [0.42, 0.15, 0.38, 0.08, 0.55],
)

XLSX.writetable("sales.xlsx", "Sales" => sales)
```

## Apply Conditional Formatting Rules

- **Sales > 4000** → green fill (good performers)
- **Sales < 1000** → red fill (underperformers)
- **Margin < 0.2** → orange fill (low margin warning)

```julia
XLSX.openxlsx("sales.xlsx"; mode="rw") do xf
    ws = xf["Sales"]

    # Style header row
    for col in ["A", "B", "C"]
        XLSX.setFont(ws, col * "1"; bold=true, color="FFFFFF")
        XLSX.setFill(ws, col * "1"; fgColor="404040")
    end

    # Green: high sales
    XLSX.setConditionalFormat(ws, "B2:B6";
        rule=">4000", fgColor="C6EFCE", fontColor="276221")

    # Red: low sales
    XLSX.setConditionalFormat(ws, "B2:B6";
        rule="<1000", fgColor="FFC7CE", fontColor="9C0006")

    # Orange: low margin
    XLSX.setConditionalFormat(ws, "C2:C6";
        rule="<0.2", fgColor="FFEB9C", fontColor="9C5700")
end
```

## Summary
- `setConditionalFormat(ws, range; rule, fgColor, fontColor)` adds a highlight rule.
- Multiple rules can apply to the same range — Excel evaluates them in order.
- Rules use Excel comparison syntax: `">4000"`, `"<0.2"`, `"=\"Error\""`.
- Colours are 6-digit hex strings without `#`.
