# 3D Contour

A 3D contour lifts contours onto a three-dimensional axis: contours of a 2D field raised by their level (`contour3d`), or **isosurfaces** of a 3D field (`contour` on an `Axis3`) — shells of constant value, the 3D analogue of a contour line. These are **interactive**, so use [GLMakie](https://github.com/MakieOrg/Makie.jl).

```julia
Pollis.packages("GLMakie")
```

```julia
using GLMakie
```

## Shared Grid, 2D and 3D Fields

```julia
r      = range(-pi, pi; length = 41)
data2d = [cos(x) + cos(y)          for x in r, y in r]
data3d = [cos(x) + cos(y) + cos(z) for x in r, y in r, z in r];
```

## Two Views Side by Side

Left: **isosurfaces** of the 3D field. Right: contours of the 2D field **lifted** into 3D by their level.

```julia
fig = Figure(size = (800, 400))
contour(fig[1, 1], -pi .. pi, -pi .. pi, -pi .. pi, data3d; axis = (type = Axis3, title = "isosurfaces"))
contour3d(fig[1, 2], r, r, data2d; levels = 10, linewidth = 3, axis = (type = Axis3, title = "lifted contours"))
fig
```

Isosurfaces are shells where the field equals a chosen level; their spacing shows how fast the field changes through the volume. **Rotate** the axes (GLMakie) to read the depth, and use transparency so inner shells show through.

## Summary
- `contour` on an `Axis3` draws isosurfaces of `f(x, y, z)`.
- `contour3d` lifts contours of `f(x, y)` to their level.
- Needs GLMakie; rotate to disambiguate depth.
