# Options Chains

Fetch the options chain for a ticker with [YFinance.jl](https://github.com/eohne/YFinance.jl) and inspect calls and puts.

```julia
using YFinance
```

## Download the Chain

`get_Options` returns a dict with `"calls"` and `"puts"`. Each is a dict of column vectors aligned by row.

```julia
op = get_Options("AMD")
calls = op["calls"]
puts  = op["puts"]
println("Columns: ", keys(calls))
```

## List Call Strikes and Prices

```julia
for i in eachindex(calls["strike"])
    println("CALL ", calls["strike"][i], "  last=", calls["lastPrice"][i])
end
```

## Most Liquid Calls by Open Interest

```julia
oi   = calls["openInterest"]
keep = sortperm(oi; rev=true)[1:min(5, length(oi))]
for i in keep
    println(calls["contractSymbol"][i], "  OI=", oi[i], "  strike=", calls["strike"][i])
end
```

## Summary
- `get_Options(symbol)` returns the nearest-expiry chain as `"calls"` and `"puts"`.
- Columns include `strike`, `lastPrice`, `bid`, `ask`, `volume`, and `openInterest`.
- High `openInterest` marks the more liquid contracts.
