# Box and Violin

Box plots give a robust five-number summary; violin plots add a kernel density so the full distribution shape — including multiple modes — is visible. Both excel at comparing groups.

```julia
using StatsPlots
using Random

Random.seed!(3);
```

## Box Plot

Three groups with different locations and spreads.

```julia
groups = repeat(["A", "B", "C"], inner = 100)
vals = [randn(100); 1.5 .+ 1.3 .* randn(100); -1 .+ 0.6 .* randn(100)]

boxplot(groups, vals; xlabel = "Group", ylabel = "Value",
    title = "Box Plot", legend = false)
```

The box spans the interquartile range, the line is the median, whiskers reach 1.5 x IQR, and points beyond are outliers.

## Violin Plot

A violin reveals shape a box plot hides. Group B below is bimodal — invisible in a box, obvious in a violin.

```julia
vals_b = [randn(100); 4 .+ randn(100)]   # bimodal
g2 = repeat(["A", "B"], inner = [100, 200])
v2 = [randn(100); vals_b]

violin(g2, v2; xlabel = "Group", ylabel = "Value", title = "Violin Plot", legend = false)
boxplot!(g2, v2; fillalpha = 0.6, width = 0.15, legend = false)
```

## Summary
- Box → robust location and spread; works for small samples.
- Violin → full shape; reveals multimodality but depends on KDE bandwidth and needs enough points.
- Overlay a thin box inside a violin to get both at once.
