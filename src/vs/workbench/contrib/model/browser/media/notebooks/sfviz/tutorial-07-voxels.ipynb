# Voxels

A voxel plot draws each cell of a **3D array grid** as a small cube, coloured by its value — the volumetric analogue of a heatmap. Best for discrete occupancy or label data. Requires [GLMakie](https://github.com/MakieOrg/Makie.jl) (GPU/display).

```julia
using GLMakie
```

## A 3D Grid

`chunk[i, j, k]` is the value of cell `(i, j, k)`. Here a 3x3x3 block labelled 1..27.

```julia
chunk = reshape(collect(1:27), 3, 3, 3);
```

## Voxels

Each cell becomes a cube coloured by its value. `gap` shrinks the cubes so you can see the separation between cells; rotate the GLMakie window to inspect interior cells.

```julia
voxels(chunk; gap = 0.33)
```

## Summary
- One cube per grid cell, coloured by value.
- `gap` separates the cubes; rotate to see inside.
- GLMakie only; for a smooth field use a volume plot instead.
