# Your First FRED Series

Download an economic time series from [FRED](https://fred.stlouisfed.org) with [FredData.jl](https://github.com/micahjsmith/FredData.jl).

**API key:** `Fred()` reads `ENV["FRED_API_KEY"]` (or `~/.freddatarc`). Set it for this session before the first call, or use the webview's *Set API key* link, which injects it into the Julia REPL.

```julia
# ENV["FRED_API_KEY"] = "YOURKEY"  # uncomment and set if not already in the environment
using FredData

f = Fred()
```

## Download Real GDP

Every series has a short **ID**. Real Gross Domestic Product is `GDPC1`.

```julia
gdp = get_data(f, "GDPC1")
println(gdp.title, " — ", gdp.units)
println(gdp.frequency, ", ", gdp.seasonal_adjustment)
```

## Inspect the Data

`get_data` returns a `FredSeries`; its `.data` field is a `DataFrame` with `date` and `value` columns.

```julia
first(gdp.data, 5)
```

```julia
last(gdp.data, 5)
```

## Summary
- `Fred()` creates a client from `FRED_API_KEY` (or `~/.freddatarc`).
- `get_data(f, "ID")` downloads a series by its FRED ID.
- The result is a `FredSeries`; read metadata fields like `title` and `units`, and the `.data` DataFrame for the observations.
