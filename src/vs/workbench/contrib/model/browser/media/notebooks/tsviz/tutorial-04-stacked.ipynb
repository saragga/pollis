# Stacked Area

A stacked area chart plots several non-negative series on top of one another: each band is one component and the upper edge is their running total. It shows the whole and each part's contribution over time. Uses StatsPlots.jl.

```julia
Pollis.packages("StatsPlots")
```

```julia
using StatsPlots, Random
```

## Component Series

Three non-negative, growing components (e.g. revenue by segment). `areaplot` stacks the columns of the matrix `Y`.

```julia
Random.seed!(42)
n = 40
t = 1:n
Y = ones(n, 3) .+ cumsum(rand(n, 3) .* 0.6; dims = 1);
```

## Stack the Areas

Each band's thickness is that component's value; the top edge is the total.

```julia
areaplot(t, Y; fillalpha = 0.6, lw = 1,
         xlabel = "Time", ylabel = "Value",
         label = ["A" "B" "C"], title = "Stacked Area")
```

Order the bands so the baseline is steady (largest or most stable at the bottom). Only the bottom band sits on a flat baseline, so upper bands are harder to read precisely — for exact per-series comparison use overlaid lines instead.

## Summary
- `areaplot(t, Y)` stacks the columns of `Y` into bands.
- Valid only for non-negative components that sum to a meaningful whole.
- Band thickness = value; top edge = total. Keep to a handful of bands.
