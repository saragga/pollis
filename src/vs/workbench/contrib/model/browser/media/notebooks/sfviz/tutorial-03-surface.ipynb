# Surface

A surface plot renders `f(x, y)` as a **height map**: the value becomes the z-coordinate, producing a rotatable 3D surface. It gives the most immediate intuition for peaks, valleys, and saddles. Interactive, so use [GLMakie](https://github.com/MakieOrg/Makie.jl).

```julia
Pollis.packages("GLMakie")
```

```julia
using GLMakie
```

## A Height Field

```julia
xs = LinRange(0, 10, 100)
ys = LinRange(0, 15, 100)
zs = [cos(x) * sin(y) for x in xs, y in ys];
```

## Render It

`axis = (type = Axis3,)` puts it on a 3D axis you can rotate and zoom.

```julia
surface(xs, ys, zs; axis = (type = Axis3,), colormap = :viridis)
```

Height is value, so the geometry *is* the field. The surface can occlude parts of itself — rotate it, and when you need exact values pair it with a contour.

## Surface plus Contour

Project contours under the surface for the best of both: intuition from the surface, precision from the lines.

```julia
fig = Figure()
ax = Axis3(fig[1, 1])
surface!(ax, xs, ys, zs; colormap = :viridis, transparency = true, alpha = 0.8)
contour3d!(ax, xs, ys, zs; levels = 12, color = :black, linewidth = 1)
fig
```

## Summary
- Surface maps `f(x, y)` to height; geometry is the field.
- Rotate (GLMakie) to overcome self-occlusion.
- Add contours when exact values matter.
