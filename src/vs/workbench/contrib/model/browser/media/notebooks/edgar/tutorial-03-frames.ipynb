# Cross-Filer Frames

The **frames** endpoint returns one concept for one period across **every** filer that reported it — a cross-section instead of a time series. Use it to rank or compare companies.

```julia
# import Pkg; Pkg.add(url = "https://github.com/Trumpingtons/EDGAR.jl")
using EDGAR

# Set your contact once via "Set contact" on the webview; it is injected as
# ENV["SEC_USER_AGENT"], which EDGAR.jl reads automatically. Or set it explicitly:
# set_user_agent("Your Name your.email@example.com")
```

## Fetch One Period

The URL ends in a **period frame**. A trailing `I` means *instantaneous* (a balance-sheet point in time); omit it for *durations* (flows over a period).

```julia
# Total assets at the end of Q4 2022, across all filers
frame = xbrl_frames("us-gaap", "Assets", "USD", "CY2022Q4I")

println("Concept: ", frame.tag, " (", frame.uom, ")")
println("Filers reporting: ", length(frame.data))
```

## Rank the Largest Filers

Each entry in `frame.data` has `cik`, `entityName`, and `val`. Sort to find the biggest.

```julia
top = sort(collect(frame.data); by = d -> d.val, rev = true)

for d in first(top, 10)
    println(rpad(d.entityName, 32), d.val)
end
```

## Duration vs Instant

For a **flow** such as revenue, use a duration frame (no trailing `I`): `CY2022` is the whole year, `CY2022Q4` just the quarter.

```julia
rev = xbrl_frames("us-gaap", "Revenues", "USD", "CY2022")
println("Filers with full-year 2022 revenue: ", length(rev.data))
```

## Summary
- Frames give one concept, one period, across all filers (a cross-section).
- Use instant frames (`...I`) for stocks (Assets, Cash) and duration frames for flows (Revenues, NetIncome).
- Each entry has `cik`, `entityName`, `val` — sort to rank, or join several frames to compare concepts.
