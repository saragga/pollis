# Histogram

A histogram reveals the shape of a distribution by binning one numeric variable. The key choice is the **bin width**.

```julia
using StatsPlots
using Random

Random.seed!(2);
```

## Basic Histogram

We draw a sample from a skewed distribution and bin it.

```julia
x = exp.(0.5 .* randn(1000))   # log-normal: right-skewed

histogram(x; bins = 30, xlabel = "Value", ylabel = "Count",
    title = "Histogram (30 bins)", label = "x")
```

## Bin Width Matters

The same data with too few and too many bins looks very different. Trust only features that survive several binnings.

```julia
p1 = histogram(x; bins = 8,  title = "8 bins",  label = "")
p2 = histogram(x; bins = 30, title = "30 bins", label = "")
p3 = histogram(x; bins = 120, title = "120 bins", label = "")
plot(p1, p2, p3; layout = (1, 3), size = (900, 280))
```

## Density Overlay

A kernel density estimate is a smooth, bin-free view. Use `normalize = :pdf` so the histogram and density share a scale.

```julia
histogram(x; bins = 30, normalize = :pdf, alpha = 0.4, label = "Histogram",
    xlabel = "Value", ylabel = "Density")
density!(x; lw = 2, label = "KDE")
```

## Summary
- Histograms show centre, spread, skew, and modality.
- Always try several bin widths.
- A density overlay avoids the binning choice entirely.
