# OHLC

An OHLC chart summarises a financial series one period at a time, encoding **open, high, low, close** as a low-to-high bar with a left tick for the open and a right tick for the close. Plots.jl has this built in via the `OHLC` type and the `ohlc` series.

```julia
using Plots, Random
```

## Simulate OHLC Bars

From a random-walk close, derive an open near it and a high/low bracketing the body.

```julia
Random.seed!(42)
n = 30
close = 100 .+ cumsum(randn(n))
open  = close .+ 0.5 .* randn(n)
high  = max.(open, close) .+ rand(n)
low   = min.(open, close) .- rand(n);
```

## Build OHLC Values

Each period is an `OHLC(open, high, low, close)` (the tuple is converted automatically).

```julia
bars = OHLC[(open[i], high[i], low[i], close[i]) for i in 1:n];
```

## Plot

The `ohlc` series draws each bar as a low-high segment with a left tick (open) and right tick (close).

```julia
ohlc(bars; xlabel = "Day", ylabel = "Price", legend = false, title = "OHLC Chart")
```

A close tick above the open tick means the period rose; below means it fell. The built-in series is single-colour — for green/red filled candlestick bodies you would draw them manually or use a finance plotting package.

## Summary
- Plots has native OHLC: the `OHLC` type and the `ohlc` series.
- Each bar: vertical low-high range, left tick = open, right tick = close.
- Single-colour; needs genuine OHLC data, not just a close series.
