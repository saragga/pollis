# Corner Plot

A corner plot shows every pairwise relationship among several variables with marginal densities on the diagonal — the standard view for posterior samples. This tutorial uses [PairPlots.jl](https://github.com/sefffal/PairPlots.jl).

```julia
Pollis.packages("PairPlots", "CairoMakie")
```

```julia
using PairPlots
using CairoMakie
using Random

Random.seed!(20);
```

## Build Correlated Samples

We simulate three variables with built-in correlation, standing in for MCMC posterior draws.

```julia
n = 2000
a = randn(n)
b = 0.7 .* a .+ 0.7 .* randn(n)
c = -0.4 .* a .+ 0.9 .* randn(n)
table = (a = a, b = b, c = c);
```

## Draw the Corner Plot

`pairplot` lays out the triangular matrix: marginals on the diagonal, pairwise densities below.

```julia
pairplot(table)
```

- The `a`–`b` panel tilts upward → positive correlation.
- The `a`–`c` panel tilts downward → negative correlation.
- Tight diagonal peaks → well-constrained variables.

## Summary
- Diagonal → marginals; off-diagonal → joint structure.
- Tilted panels reveal correlation; banana shapes warn of weak identifiability.
- Ideal for posterior samples and moderate variable sets.
