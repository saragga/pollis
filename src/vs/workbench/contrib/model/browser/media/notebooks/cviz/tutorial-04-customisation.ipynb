# Customisation

Themes, colours, annotations, layouts, and saving publication-quality figures with [Plots.jl](https://github.com/JuliaPlots/Plots.jl).

```julia
Pollis.packages("Plots")
```

```julia
using Plots
using Random

Random.seed!(4);
```

## Themes

A theme sets fonts, colours, and gridlines consistently. Apply one globally with `theme`.

```julia
theme(:default)
x = 1:50
y = cumsum(randn(50))
plot(x, y; lw = 2, label = "series", title = "Default Theme")
```

## Colours and Annotations

Set a palette, label axes, and annotate a point of interest.

```julia
plt = plot(x, y; lw = 2.5, color = :purple, label = "series",
    xlabel = "Step", ylabel = "Value", title = "Annotated", legend = :topleft)
imax = argmax(y)
scatter!([x[imax]], [y[imax]]; color = :red, label = "peak")
annotate!(x[imax], y[imax], text("  peak", :left, 9))
```

## Layouts

Combine several panels with `layout`.

```julia
p1 = plot(x, y; title = "A", legend = false)
p2 = histogram(randn(500); title = "B", legend = false)
p3 = scatter(randn(100), randn(100); title = "C", legend = false)
plot(p1, p2, p3; layout = (1, 3), size = (900, 280))
```
