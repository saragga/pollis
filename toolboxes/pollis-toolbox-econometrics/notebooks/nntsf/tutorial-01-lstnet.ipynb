# LSTnet — Neural Network Time-Series Forecasting

Long- and Short-term Time-series network (Lai et al., 2018): a CNN for short-term local patterns, a recurrent-skip GRU for long seasonal cycles, plus an autoregressive term.

**Not written yet.** This is a starter stub — run the cell below to train the model on the bundled sample data.

```julia
using FluxArchitectures

poollength = 10
input, target = get_data(:exchange_rate, poollength, 1000, 6)
inputsize = size(input, 1)

model = LSTnet(inputsize, 2, 3, poollength, 120)
loss(x, y) = (Flux.reset!(model); Flux.mse(model(x), transpose(y)))
opt = Flux.ADAM(0.007)
Flux.train!(loss, Flux.params(model), Iterators.repeated((input, target), 20), opt)
```
