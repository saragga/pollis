# Adaptive and Improper Integrals

Adaptive Gauss-Kronrod quadrature (QuadGK) splits the interval where its error estimate is largest. It copes with peaks, endpoint singularities and infinite limits, and it reports how accurate the answer is.

```julia
Pollis.packages("QuadGK")
```

```julia
using QuadGK
```

## A sharp peak

```julia
peak(x) = 1 / (1e-4 + (x - 0.3)^2)
I, E, n = quadgk_count(peak, 0, 1; rtol=1e-10)
(integral = I, error_estimate = E, evaluations = n, exact = 100 * (atan(70) + atan(30)))
```

## Endpoint singularities

$\int_0^1 x^{-1/2}\,dx = 2$. QuadGK never evaluates at the endpoints, so the singularity is harmless.

```julia
quadgk_count(x -> 1 / sqrt(x), 0, 1)
```

## Infinite limits

A change of variables maps $[0, \infty)$ to a finite interval. The normal density must integrate to 1, and the exponential distribution's mean is $1/\lambda$.

```julia
normal(x) = exp(-x^2 / 2) / sqrt(2pi)
lambda = 0.5
(normal_total = quadgk(normal, -Inf, Inf)[1], exponential_mean = quadgk(x -> x * lambda * exp(-lambda * x), 0, Inf)[1])
```

## Kinks: tell QuadGK where they are

Extra points split the interval, so each piece is smooth.

```julia
kink(x) = abs(x - 0.7)
(without_split = quadgk_count(kink, 0, 1), with_split = quadgk_count(kink, 0, 0.7, 1))
```
