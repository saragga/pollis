# From Riemann Sums to the Fundamental Theorem

The integral is the limit of sums of thin rectangles. The running total $F(x) = \int_a^x f$ has slope $f$: that is the fundamental theorem of calculus, and it is why antiderivatives evaluate integrals.

```julia
using Plots
f(x) = 3 + 2sin(x) + 0.3x
a, b = 0.0, 6.0
```

## Left, right and midpoint sums

```julia
function riemann(f, a, b, n; point=:mid)
    h = (b - a) / n
    xs = point == :left ? (a .+ (0:n-1) .* h) : point == :right ? (a .+ (1:n) .* h) : (a .+ ((1:n) .- 0.5) .* h)
    return h * sum(f, xs)
end
exact = 3(b - a) - 2(cos(b) - cos(a)) + 0.15(b^2 - a^2)
[(n = n, left = riemann(f, a, b, n; point=:left), mid = riemann(f, a, b, n), right = riemann(f, a, b, n; point=:right), exact = exact) for n in (4, 16, 64, 256)]
```

```julia
n = 8; h = (b - a) / n; mids = a .+ ((1:n) .- 0.5) .* h
xs = range(a, b, length=300)
bar(mids, f.(mids), bar_width=h, alpha=0.3, label="midpoint rectangles, n = $n")
plot!(xs, f.(xs), lw=2, c=:black, label="f(x)", xlabel="x")
```

## The accumulation function

$F(x) = \int_a^x f(t)\,dt$. Its derivative should equal $f$.

```julia
using QuadGK
F(x) = quadgk(f, a, x)[1]
xs = range(a, b, length=200)
slope = [(F(x + 1e-5) - F(x - 1e-5)) / 2e-5 for x in xs[2:end-1]]
plot(xs, F.(xs), lw=2, label="F(x)", layout=(2, 1), xlabel="x")
plot!(xs[2:end-1], [slope f.(xs[2:end-1])], subplot=2, lw=2, ls=[:solid :dash], label=["slope of F" "f"])
```
